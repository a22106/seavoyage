```python
import sys
sys.path.append("..")

import networkx as nx
from seavoyage.classes.m_network import MNetwork
import seavoyage as sv

G = nx.Graph()
G.add_node(1, x=129.0, y=35.0)
G.add_node(2, x=130.0, y=36.0)
G.add_edge(1, 2, weight=1.5)

# MNetwork로 변환
mnet = MNetwork.from_networkx(G)
```

```python
mnet.to_geojson()
```

```python
print(mnet.nodes(data=True))
print(mnet.edges(data=True))
```
