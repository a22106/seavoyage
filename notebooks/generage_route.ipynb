```python
#!/usr/bin/env python
"""
Seavoyage vs Haversine 직접 비교 테스트
seavoyage 라이브러리로 직접 경로를 생성하고 haversine 패키지로 계산한 값과 비교
"""

import sys
import os
import numpy as np
from typing import List

import seavoyage as sv
from haversine import haversine, Unit


# 테스트 케이스 정의
TEST_ROUTES = [
    {
        "name": "북태평양 경로 (알래스카-시애틀)",
        "points": [
            [-146.32237259397868, 59.25525367185301],  # 알래스카
            [-135.85873254042957, 56.679835558483035],  # 브리티시컬럼비아
            [-127.0300362452476, 47.66770310902919],   # 시애틀
        ],
        "units": "nm"
    }
]

def calculate_haversine_distance(coordinates: List[List[float]], units: str = 'nm') -> float:
    """
    좌표 리스트를 따라 haversine 공식으로 총 거리를 계산
    """
    if not coordinates or len(coordinates) < 2:
        return 0.0
    
    # 단위 매핑
    unit_map = {
        'nm': Unit.NAUTICAL_MILES,
        'km': Unit.KILOMETERS,
        'm': Unit.METERS,
        'mi': Unit.MILES
    }
    
    haversine_unit = unit_map.get(units, Unit.NAUTICAL_MILES)
    
    total_distance = 0.0
    for i in range(len(coordinates) - 1):
        # 좌표 변환 (경도, 위도 -> 위도, 경도)
        point1 = (coordinates[i][1], coordinates[i][0])
        point2 = (coordinates[i+1][1], coordinates[i+1][0])
        segment_distance = haversine(point1, point2, unit=haversine_unit)
        total_distance += segment_distance
    
    return total_distance

def generate_seavoyage_routes(test_case: dict) -> List[dict]:
    """
    seavoyage를 사용해서 경로를 생성
    """
    points = test_case["points"]
    units = test_case["units"]
    
    routes = []
    for i in range(len(points) - 1):
        try:
            # seavoyage로 경로 생성
            route = sv.seavoyage(
                points[i], 
                points[i + 1], 
                units=units,
            )
            
            routes.append({
                "from": points[i],
                "to": points[i + 1],
                "route": route,
                "segment_index": i + 1
            })
            
        except Exception as e:
            print(f"⚠️  경로 생성 실패 {points[i]} -> {points[i+1]}: {e}")
            # 실패한 경우 직선 거리로 대체
            direct_distance = haversine(
                (points[i][1], points[i][0]), 
                (points[i+1][1], points[i+1][0]), 
                unit=Unit.NAUTICAL_MILES if units == 'nm' else Unit.KILOMETERS
            )
            
            routes.append({
                "from": points[i],
                "to": points[i + 1],
                "route": {
                    "geometry": {
                        "coordinates": [points[i], points[i + 1]]
                    },
                    "properties": {
                        "length": direct_distance,
                        "units": units
                    }
                },
                "segment_index": i + 1,
                "is_fallback": True
            })
    
    return routes

def analyze_route_comparison(test_case: dict) -> dict:
    """
    단일 테스트 케이스에 대한 비교 분석
    """
    print(f"\n{'='*80}")
    print(f"🗺️  테스트: {test_case['name']}")
    print(f"{'='*80}")
    
    # seavoyage로 경로 생성
    routes = generate_seavoyage_routes(test_case)
    
    total_seavoyage_length = 0
    total_haversine_length = 0
    segment_comparisons = []
    
    for route_info in routes:
        segment_idx = route_info["segment_index"]
        from_point = route_info["from"]
        to_point = route_info["to"]
        route = route_info["route"]
        is_fallback = route_info.get("is_fallback", False)
        
        print(f"\n📍 구간 {segment_idx}: {from_point} -> {to_point}")
        
        # seavoyage 결과
        seavoyage_length = route["properties"]["length"]
        coordinates = route["geometry"]["coordinates"]
        
        print(f"  🚢 Seavoyage 길이: {seavoyage_length:.2f} {test_case['units']}")
        print(f"  📏 경로 포인트 수: {len(coordinates)}")
        
        if is_fallback:
            print(f"  ⚠️  직선 거리로 대체됨")
        
        # haversine 계산
        haversine_length = calculate_haversine_distance(coordinates, test_case['units'])
        
        print(f"  🧮 Haversine 길이: {haversine_length:.2f} {test_case['units']}")
        
        # 직선 거리 계산
        direct_length = haversine(
            (from_point[1], from_point[0]),
            (to_point[1], to_point[0]),
            unit=Unit.NAUTICAL_MILES if test_case['units'] == 'nm' else Unit.KILOMETERS
        )
        print(f"  ➡️  직선 거리: {direct_length:.2f} {test_case['units']}")
        
        # 차이 계산
        sv_vs_haversine = abs(seavoyage_length - haversine_length)
        sv_vs_direct = abs(seavoyage_length - direct_length)
        haversine_vs_direct = abs(haversine_length - direct_length)
        
        print(f"  📊 비교:")
        print(f"     Seavoyage vs Haversine: {sv_vs_haversine:.2f} {test_case['units']} ({(sv_vs_haversine/seavoyage_length)*100:.1f}%)")
        print(f"     Seavoyage vs 직선: {sv_vs_direct:.2f} {test_case['units']} ({(sv_vs_direct/seavoyage_length)*100:.1f}%)")
        print(f"     Haversine vs 직선: {haversine_vs_direct:.2f} {test_case['units']} ({(haversine_vs_direct/haversine_length)*100:.1f}%)")
        
        # 데이터 수집
        segment_comparisons.append({
            "segment": segment_idx,
            "seavoyage_length": seavoyage_length,
            "haversine_length": haversine_length,
            "direct_length": direct_length,
            "sv_vs_haversine_diff": sv_vs_haversine,
            "sv_vs_haversine_pct": (sv_vs_haversine/seavoyage_length)*100,
            "coordinates_count": len(coordinates),
            "is_fallback": is_fallback
        })
        
        total_seavoyage_length += seavoyage_length
        total_haversine_length += haversine_length
    
    # 전체 결과 요약
    print(f"\n🎯 전체 경로 요약:")
    print(f"  총 Seavoyage 길이: {total_seavoyage_length:.2f} {test_case['units']}")
    print(f"  총 Haversine 길이: {total_haversine_length:.2f} {test_case['units']}")
    
    total_diff = abs(total_seavoyage_length - total_haversine_length)
    total_diff_pct = (total_diff/total_seavoyage_length)*100
    
    print(f"  총 차이: {total_diff:.2f} {test_case['units']} ({total_diff_pct:.1f}%)")
    
    if total_diff_pct < 1.0:
        print(f"  ✅ 매우 정확함 (1% 이내)")
    elif total_diff_pct < 5.0:
        print(f"  ✅ 정확함 (5% 이내)")
    elif total_diff_pct < 10.0:
        print(f"  ⚠️  보통 (10% 이내)")
    else:
        print(f"  ❌ 큰 차이 (10% 초과)")
    
    return {
        "test_name": test_case["name"],
        "total_seavoyage_length": total_seavoyage_length,
        "total_haversine_length": total_haversine_length,
        "total_diff": total_diff,
        "total_diff_pct": total_diff_pct,
        "segment_count": len(segment_comparisons),
        "segments": segment_comparisons,
        "units": test_case["units"]
    }

def run_all_tests():
    """
    모든 테스트 케이스 실행
    """
    print("🚢 Seavoyage vs Haversine 비교 테스트 시작")
    print("="*100)
    
    all_results = []
    
    for test_case in TEST_ROUTES:
        try:
            result = analyze_route_comparison(test_case)
            all_results.append(result)
        except Exception as e:
            print(f"❌ 테스트 실패 '{test_case['name']}': {e}")
            continue
    
    # 전체 통계 분석
    print(f"\n{'='*100}")
    print("📈 전체 테스트 결과 통계")
    print(f"{'='*100}")
    
    if all_results:
        # 차이 백분율 통계
        diff_pcts = [r["total_diff_pct"] for r in all_results]
        
        print(f"📊 정확도 통계:")
        print(f"  평균 차이: {np.mean(diff_pcts):.2f}%")
        print(f"  최소 차이: {np.min(diff_pcts):.2f}%")
        print(f"  최대 차이: {np.max(diff_pcts):.2f}%")
        print(f"  표준편차: {np.std(diff_pcts):.2f}%")
        
        # 정확도 등급별 분류
        very_accurate = sum(1 for pct in diff_pcts if pct < 1.0)
        accurate = sum(1 for pct in diff_pcts if 1.0 <= pct < 5.0)
        moderate = sum(1 for pct in diff_pcts if 5.0 <= pct < 10.0)
        poor = sum(1 for pct in diff_pcts if pct >= 10.0)
        
        print(f"\n🎯 정확도 분포:")
        print(f"  매우 정확 (1% 이내): {very_accurate}개")
        print(f"  정확 (1-5%): {accurate}개") 
        print(f"  보통 (5-10%): {moderate}개")
        print(f"  부정확 (10% 초과): {poor}개")
        
        # 최고/최악 케이스
        best_idx = np.argmin(diff_pcts)
        worst_idx = np.argmax(diff_pcts)
        
        print(f"\n🏆 가장 정확한 경로: {all_results[best_idx]['test_name']} ({diff_pcts[best_idx]:.2f}%)")
        print(f"📉 가장 부정확한 경로: {all_results[worst_idx]['test_name']} ({diff_pcts[worst_idx]:.2f}%)")
    
    # 결론
    print(f"\n{'='*100}")
    print("💡 결론")
    print(f"{'='*100}")
    
    if np.mean(diff_pcts) < 5.0:
        print("✅ Haversine 계산이 Seavoyage 결과와 전반적으로 잘 일치합니다.")
        print("   API에서 Haversine을 사용하는 것이 적절합니다.")
    else:
        print("⚠️  Haversine과 Seavoyage 사이에 상당한 차이가 있습니다.")
        print("   추가 분석이 필요할 수 있습니다.")
    
    return all_results

if __name__ == "__main__":
    results = run_all_tests() 
```
