```python
import sys
sys.path.append("..")
import seavoyage as sv


start_point = (132.0, 34.3)  # JWC 서쪽 경계 바로 옆
end_point = (136.0, 34.3)    # JWC 동쪽 경계 바로 옆

def test_route(start_point, end_point):
    normal_route = sv.seavoyage(
        start_point, 
        end_point
    )

    restricted_route = sv.seavoyage(
        start_point, 
        end_point,
        restrictions=['jwc']
    )

    print("Normal route:", normal_route)
    print("Restricted route:", restricted_route)
    return normal_route, restricted_route

test_route(start_point, end_point)
sv.register_custom_restriction('jwc', "./restrictions/jwc.geojson")
normal_route, restrict_route = test_route(start_point, end_point)
```

```python
import folium

def generate_route_map(route):
    """경로를 표시하는 Folium 맵 생성"""
    # 출발지와 목적지 좌표 추출
    m = folium.Map(location=[34.3, 134.3], zoom_start=10)
    m = folium.GeoJson(route).add_to(m)
    return m
    
map_normal = generate_route_map(normal_route)
map_normal.save("normal_route.html")
map_restricted = generate_route_map(restrict_route)
map_restricted.save("restricted_route.html")
```
