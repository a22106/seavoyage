```python
import sys
sys.path.append("..")
import seavoyage as sv

start_point = (132.0, 34.3)  # JWC 서쪽 경계 바로 옆
end_point = (136.0, 34.3)    # JWC 동쪽 경계 바로 옆

normal_route = sv.seavoyage(
    start_point, 
    end_point
)

restricted_route = sv.seavoyage(
    start_point, 
    end_point,
    restrictions=['jwc']
)
```

```python
import folium

def generate_route_map(route):
    """경로를 표시하는 Folium 맵 생성"""
    # 출발지와 목적지 좌표 추출
    m = folium.Map(location=[34.3, 134.3], zoom_start=10)
    m = folium.GeoJson(route).add_to(m)
    return m
    
map = generate_route_map(normal_route)
map.save("normal_route.html")
```

```python
map2 = generate_route_map(restricted_route)
map2.save("restricted_route.html")
```
