```python
import seavoyage as sv
from haversine import haversine, Unit

point1 = (135.193805, 34.653909) # Kobe
point2 = (129.491573, 35.395965) # Ulsan
point3 = (126.635678, 35.993098) # Kunsan
graph = sv.MNetwork().from_geojson('test_geojson/default.geojson')

route = sv.seavoyage(point1, point2, M=graph, units='mi')
length1 = route['properties']['length']
m = sv.utils.map_folium(route)
print(f"route1 len: {length1}")

# calculate and print distance for each points in the route and print the sum of the distances
total_distance = 0
distances = []
coordinates = route['geometry']['coordinates']

for i in range(1, len(coordinates)):
    point1 = coordinates[i-1]
    point1 = (point1[1], point1[0])
    point2 = coordinates[i]
    point2 = (point2[1], point2[0])
    
    distance = haversine(point1, point2, unit=Unit.MILES)
    
    distances.append(distance)
    total_distance += distance

print(f"총 거리: {total_distance} nm")
```
