```python
import sys
sys.path.append("..")

import seavoyage as sv

sv.register_custom_restriction('hra', "./restrictions/hra.geojson")
sv.register_custom_restriction('jwc', "./restrictions/jwc.geojson")
```

```python
from seavoyage.exceptions import DestinationInRestrictionError


start_point1 = (132.0, 34.3)  
end_point1 = (57.0, 24.828) # 제한구역 내부의 도착점

def test_route(start_point, end_point):
    normal_route = sv.seavoyage(
        start_point, 
        end_point
    )

    restricted_route = sv.seavoyage(
        start_point, 
        end_point,
        restrictions=['jwc']
    )

    print("Normal route:", normal_route)
    print("Restricted route:", restricted_route)
    return normal_route, restricted_route

try:# 제한구역 내부 도착점 시 에러 발생 테스트트
    normal_route, restrict_route = test_route(start_point1, end_point1)
except DestinationInRestrictionError as e:
    print(e)
```
